```python
# ============================================================
# BLOQUE 5: CALCULADORA RECRIPEI-CR
# Evalúa el nivel de madurez de una PyME según el modelo
# ============================================================

def calcular_nivel_recripei_cr(empresa: dict) -> dict:
    """
    Calculadora del Modelo RECRIPEI-CR (Blanco, 2018 + Druva, 2025)
    Evalúa el nivel de madurez de ciberresiliencia de una PyME.
    
    Parámetros:
    -----------
    empresa : dict con las siguientes claves:
        - tiene_backup_inmutable (bool)
        - backup_aislado (bool) 
        - tiene_mfa (bool)
        - tiene_cifrado (bool)
        - tiene_sandbox (bool)
        - verifica_backups (bool)
        - tiene_logs (bool)
        - analisis_forense (bool)
        - monitoreo_continuo (bool)
        - coordina_incibe (bool)
    
    Retorna:
    --------
    dict con nivel, nombre, descripción, normativa y recomendaciones
    """
    
    puntos = 0
    
    # Nivel 1 - Área Restringida: Inmutabilidad
    n1 = empresa.get('tiene_backup_inmutable', False) and empresa.get('backup_aislado', False)
    if n1: puntos += 1
    
    # Nivel 2 - Área Crítica: Seguridad en copias
    n2 = empresa.get('tiene_mfa', False) and empresa.get('tiene_cifrado', False)
    if n2 and n1: puntos += 1
    
    # Nivel 3 - Área Protegida: Remediación
    n3 = empresa.get('tiene_sandbox', False) and empresa.get('verifica_backups', False)
    if n3 and puntos >= 2: puntos += 1
    
    # Nivel 4 - Área de Exclusión: Investigación
    n4 = empresa.get('tiene_logs', False) and empresa.get('analisis_forense', False)
    if n4 and puntos >= 3: puntos += 1
    
    # Nivel 5 - Área de Influencia: Detección mejorada
    n5 = empresa.get('monitoreo_continuo', False) and empresa.get('coordina_incibe', False)
    if n5 and puntos >= 4: puntos += 1
    
    # Definición de niveles
    niveles = {
        0: {
            'nivel': 0,
            'nombre': '⛔ Sin protección',
            'area_recripei': 'Sin área asignada',
            'descripcion': 'La empresa no dispone de medidas básicas de ciberresiliencia.',
            'normativa': 'Incumple art. 32 RGPD y art. 21 NIS2',
            'riesgo': 'CRÍTICO',
            'rto': 'Indefinido',
            'rpo': 'Pérdida total posible',
            'recomendaciones': [
                '1. Implementar backups inmutables inmediatamente (Azure Backup / OneDrive)',
                '2. Activar MFA en todas las cuentas',
                '3. Contactar con INCIBE (017) para asesoramiento gratuito'
            ]
        },
        1: {
            'nivel': 1,
            'nombre': '🟡 Nivel 1 — Inmutabilidad de datos',
            'area_recripei': 'Área Restringida',
            'descripcion': 'La empresa dispone de copias de seguridad inmutables y aisladas.',
            'normativa': 'Cumple parcialmente art. 32 RGPD y art. 21.2.c NIS2',
            'riesgo': 'ALTO',
            'rto': '24-72 horas',
            'rpo': 'Hasta 24 horas de datos',
            'recomendaciones': [
                '1. Implementar MFA en todos los accesos al sistema de backup',
                '2. Cifrar datos en tránsito y en reposo',
                '3. Revisar política de control de accesos'
            ]
        },
        2: {
            'nivel': 2,
            'nombre': '🟠 Nivel 2 — Seguridad en las copias',
            'area_recripei': 'Área Crítica',
            'descripcion': 'La empresa protege su infraestructura de backup con MFA y cifrado.',
            'normativa': 'Cumple art. 32 RGPD y art. 21 NIS2 (MFA)',
            'riesgo': 'MEDIO-ALTO',
            'rto': '12-24 horas',
            'rpo': 'Hasta 8 horas de datos',
            'recomendaciones': [
                '1. Implementar entorno sandbox para verificar backups antes de restaurar',
                '2. Establecer proceso de verificación de integridad post-restauración',
                '3. Documentar el plan de recuperación (obligatorio art. 33 RGPD)'
            ]
        },
        3: {
            'nivel': 3,
            'nombre': '🔵 Nivel 3 — Remediación cibernética',
            'area_recripei': 'Área Protegida',
            'descripcion': 'La empresa garantiza recuperaciones limpias sin riesgo de reinfección.',
            'normativa': 'Cumple art. 32 y art. 33 RGPD — recuperación en 72h',
            'riesgo': 'MEDIO',
            'rto': '4-12 horas',
            'rpo': '1-4 horas de datos',
            'recomendaciones': [
                '1. Implementar análisis de logs con Microsoft Sentinel o herramienta equivalente',
                '2. Establecer proceso de análisis forense post-incidente',
                '3. Documentar todos los incidentes para cumplimiento art. 264 CP'
            ]
        },
        4: {
            'nivel': 4,
            'nombre': '🟢 Nivel 4 — Investigación cibernética',
            'area_recripei': 'Área de Exclusión',
            'descripcion': 'La empresa analiza forensemente los incidentes y documenta las brechas.',
            'normativa': 'Cumple art. 264 CP y art. 33-34 RGPD — evidencia jurídica',
            'riesgo': 'BAJO-MEDIO',
            'rto': '2-6 horas',
            'rpo': 'Menos de 1 hora',
            'recomendaciones': [
                '1. Implementar monitoreo continuo con XDR/NDR',
                '2. Establecer protocolo de coordinación con INCIBE y AEPD',
                '3. Activar el modelo como instrumento de compliance (art. 31 bis CP)'
            ]
        },
        5: {
            'nivel': 5,
            'nombre': '🏆 Nivel 5 — Detección mejorada',
            'area_recripei': 'Área de Influencia',
            'descripcion': 'Estándar de oro: supervisión proactiva y coordinación externa.',
            'normativa': 'Cumple art. 31 bis CP — eximente de responsabilidad penal corporativa',
            'riesgo': 'MÍNIMO',
            'rto': 'Minutos - horas',
            'rpo': 'Casi cero',
            'recomendaciones': [
                '✅ Empresa con máxima madurez de ciberresiliencia',
                '✅ Posición jurídica sólida ante tribunal y AEPD',
                '📋 Mantener auditorías periódicas y actualización del modelo'
            ]
        }
    }
    
    return niveles[puntos]


# --- EJEMPLO DE USO ---
print('=' * 60)
print('  CALCULADORA RECRIPEI-CR | DCS Consulting')
print('  Modelo de Madurez de Ciberresiliencia para PyMEs')
print('=' * 60)

# Perfil PyME tipo (gestoría, 12 empleados, Microsoft 365)
pyme_ejemplo = {
    'nombre': 'Gestoría Mediterráneo SL',
    'sector': 'Servicios administrativos',
    'empleados': 12,
    'tiene_backup_inmutable': True,   # Tiene backups pero no inmutables
    'backup_aislado': False,           # No están aislados
    'tiene_mfa': False,                # Sin MFA
    'tiene_cifrado': False,            # Sin cifrado
    'tiene_sandbox': False,
    'verifica_backups': False,
    'tiene_logs': False,
    'analisis_forense': False,
    'monitoreo_continuo': False,
    'coordina_incibe': False
}

resultado = calcular_nivel_recripei_cr(pyme_ejemplo)

print(f"\n📊 DIAGNÓSTICO: {pyme_ejemplo['nombre']}")
print(f"   Sector: {pyme_ejemplo['sector']} | Empleados: {pyme_ejemplo['empleados']}")
print(f"\n{resultado['nombre']}")
print(f"   Área RECRIPEI: {resultado['area_recripei']}")
print(f"   Descripción:   {resultado['descripcion']}")
print(f"   Normativa:     {resultado['normativa']}")
print(f"   Nivel de riesgo: {resultado['riesgo']}")
print(f"   RTO estimado:  {resultado['rto']}")
print(f"   RPO estimado:  {resultado['rpo']}")
print(f"\n📋 RECOMENDACIONES:")
for rec in resultado['recomendaciones']:
    print(f"   {rec}")
```

```python
# ============================================================
# BLOQUE 6: VISUALIZACIÓN DEL NIVEL RECRIPEI-CR
# ============================================================

def visualizar_nivel(empresa_config: dict, resultado: dict):
    fig, ax = plt.subplots(figsize=(10, 4))
    
    niveles_nombres = [
        'Nivel 1\nInmutabilidad',
        'Nivel 2\nSeguridad\ncopias',
        'Nivel 3\nRemediación',
        'Nivel 4\nInvestigación',
        'Nivel 5\nDetección\nmejorada'
    ]
    
    nivel_actual = resultado['nivel']
    
    for i, nombre in enumerate(niveles_nombres):
        nivel_num = i + 1
        if nivel_num <= nivel_actual:
            color = NAVY
            alpha = 1.0
        else:
            color = GRAY
            alpha = 0.3
        
        rect = mpatches.FancyBboxPatch(
            (i * 1.8 + 0.1, 0.3), 1.5, 0.4 + i * 0.1,
            boxstyle='round,pad=0.05',
            facecolor=color, alpha=alpha, edgecolor='white', linewidth=2
        )
        ax.add_patch(rect)
        ax.text(i * 1.8 + 0.85, 0.5 + i * 0.05, nombre,
                ha='center', va='center', fontsize=8,
                color='white', fontweight='bold')
    
    ax.set_xlim(-0.2, 9.2)
    ax.set_ylim(0, 1.2)
    ax.axis('off')
    
    ax.set_title(
        f'Nivel RECRIPEI-CR: {resultado["nombre"]} | Riesgo: {resultado["riesgo"]}',
        fontsize=12, fontweight='bold', color=NAVY
    )
    
    fig.text(0.5, 0.02,
             'Modelo RECRIPEI-CR © Blanco (2018) + Druva (2025) | DCS Consulting',
             ha='center', fontsize=8, color=GRAY, style='italic')
    
    plt.tight_layout()
    plt.savefig('output/04_nivel_recripei_cr.png', dpi=150, bbox_inches='tight')
    plt.show()
    print('✅ Gráfico 4 guardado: nivel RECRIPEI-CR')

visualizar_nivel(pyme_ejemplo, resultado)
```
